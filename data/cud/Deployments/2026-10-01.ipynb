```python
%sql
-- =============================================================================
-- TABLE 1: job_task_log  (per-table/task detail — written by every area)
-- =============================================================================
CREATE TABLE IF NOT EXISTS `p-cdh`.pipeline.job_task_log (
    pipeline_area       STRING          NOT NULL    COMMENT 'Area: staging, landing, landed, curated',
    business_date       DATE            NOT NULL    COMMENT 'Business date being processed',
    ingestion_date      DATE                        COMMENT 'Ingestion/trigger date (current_date at run time)',
    schema_name         STRING                      COMMENT 'Source schema: profits, thesis, qca, etc.',
    group_name          STRING                      COMMENT 'Group within schema (as defined in Redis)',
    task_name           STRING          NOT NULL    COMMENT 'What was processed (e.g., profits_loan, party_qca)',
    status              STRING          NOT NULL    COMMENT 'running | completed | failed',
    start_timestamp     TIMESTAMP                   COMMENT 'When this task started',
    end_timestamp       TIMESTAMP                   COMMENT 'When this task finished',
    duration_seconds    INT                         COMMENT 'Elapsed seconds',
    databricks_run_id   STRING                      COMMENT 'Databricks job run ID for this execution',
    job_name            STRING                      COMMENT 'Databricks job name',
    error_message       STRING                      COMMENT 'Error details (null if success)',
    triggered_by        STRING                      COMMENT 'Notebook name that called the function (auto-detected)',
    updated_at          TIMESTAMP       NOT NULL    COMMENT 'Last write timestamp'
)
USING DELTA
PARTITIONED BY (business_date)
COMMENT 'Analytical table — one row per table/task processed by Databricks jobs.'
TBLPROPERTIES (
    'delta.columnMapping.mode' = 'name',
    'delta.minReaderVersion' = '2',
    'delta.minWriterVersion' = '7',
    'delta.enableDeletionVectors' = 'true'
);

-- =============================================================================
-- TABLE 2: job_execution_log  (area-level completion — written when an area finishes)
-- =============================================================================
CREATE TABLE IF NOT EXISTS `p-cdh`.pipeline.job_execution_log (
    pipeline_area       STRING          NOT NULL    COMMENT 'Area: staging, landing, landed, curated',
    business_date       DATE            NOT NULL    COMMENT 'Business date that was processed',
    ingestion_date      DATE                        COMMENT 'Date the pipeline was triggered',
    status              STRING          NOT NULL    COMMENT 'running | completed | failed',
    start_timestamp     TIMESTAMP                   COMMENT 'When the area started (MIN of tasks)',
    end_timestamp       TIMESTAMP                   COMMENT 'When the area finished (MAX of tasks)',
    duration_seconds    INT                         COMMENT 'Total elapsed seconds',
    triggered_by        STRING                      COMMENT 'auto_completion_check | manual',
    databricks_run_id   STRING                      COMMENT 'Databricks job run ID',
    error_message       STRING                      COMMENT 'Error details (null if success)',
    logged_at           TIMESTAMP       NOT NULL    COMMENT 'When this record was written to the log'
)
USING DELTA
PARTITIONED BY (business_date)
COMMENT 'Historical log of pipeline area executions — one row per area per execution.'
TBLPROPERTIES (
    'delta.columnMapping.mode' = 'name',
    'delta.minReaderVersion' = '2',
    'delta.minWriterVersion' = '5'
);
```
