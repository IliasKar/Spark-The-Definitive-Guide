```python
%sql
CREATE OR REPLACE TABLE `p-cdh`.primaryarchive.exposure AS
SELECT * FROM `qa-cdh`.primaryarchive.exposure2hist;
```

```python
%sql
CREATE OR REPLACE VIEW `p-cdh`.primary.exposure AS  
SELECT
  *
FROM
  `p-cdh`.primaryarchive.exposure
WHERE
  BusinessDate < (
    SELECT
      MIN(BusinessDate) as start_date
    FROM
      `p-cdh`.primaryactive.exposure
  )
UNION ALL
SELECT
  *
FROM
  `p-cdh`.primaryactive.exposure;
```
