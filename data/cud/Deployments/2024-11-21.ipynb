## WIF-454 : Add CommonPortfolioID, CommonPortfolio in portfolio

```python

env = 'p'

# Add columns in Unity
spark.sql(f'''
    ALTER TABLE `{env}-cdh`.`primaryactive`.portfolio
    ADD COLUMNS (
        CommonPortfolioID STRING AFTER OwnerId,
        CommonPortfolio STRING AFTER CommonPortfolioID
    )
''')

# Update fields
spark.sql(f'''
            UPDATE `{env}-cdh`.`primaryactive`.portfolio
            SET 
                CommonPortfolioID = CASE 
                  WHEN PortfolioId IN (19, 20, 21) THEN 18
                  WHEN PortfolioId IN (34, 35) THEN 36
                  WHEN PortfolioId IN (47, 48, 50) THEN 49
                  END,
                CommonPortfolio = CASE 
                  WHEN PortfolioId IN (19, 20, 21) THEN 'OrionGalaxy'
                  WHEN PortfolioId IN (34, 35) THEN 'CosmosRetailWhl'
                  WHEN PortfolioId IN (47, 48, 50) THEN 'HermesABC'
                  END
          ''')
```
