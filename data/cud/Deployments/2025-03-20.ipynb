```python
%sql
CREATE TABLE `p-cdh`.primaryarchive.exposure AS
SELECT * FROM `qa-cdh`.primaryarchive.exposurehist;
```

```python
%sql
CREATE TABLE `p-cdh`.primaryarchive.party AS
SELECT * FROM `qa-cdh`.primaryarchive.partyhist;
```

```python
%sql
CREATE OR REPLACE VIEW `p-cdh`.primary.exposure AS  
SELECT
  *
FROM
  `p-cdh`.primaryarchive.exposure
WHERE
  BusinessDate < (
    SELECT
      MIN(BusinessDate) as start_date
    FROM
      `p-cdh`.primaryactive.exposure
  )
UNION ALL
SELECT
  *
FROM
  `p-cdh`.primaryactive.exposure;
```

```python
%sql
CREATE OR REPLACE VIEW `p-cdh`.primary.party AS  
SELECT
  *
FROM
  `p-cdh`.primaryarchive.party
WHERE
  BusinessDate < (
    SELECT
      MIN(BusinessDate) as start_date
    FROM
      `p-cdh`.primaryactive.party
  )
UNION ALL
SELECT
  *
FROM
  `p-cdh`.primaryactive.party;
```
