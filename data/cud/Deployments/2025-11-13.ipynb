###Closed loans function update

```python
env='p'
# Get the schema of the dexposure table
dexposure_table = spark.table(f"`{env}-cdh`.derivative.dexposure")
dexposure_schema_fields = dexposure_table.schema.fields


# Convert the schema fields into the required string format
dexposure_schema = ",\n".join([
    f"{field.name} {field.dataType.simpleString()}"
    for field in dexposure_schema_fields
])


spark.sql(f'''
CREATE OR REPLACE FUNCTION `{env}-cdh`.derivative.get_all_exposures_by_business_date(
    business_date Date
)
RETURNS TABLE 
( 
{dexposure_schema}
)
RETURN

SELECT B.*
FROM(
        SELECT  ExposureKey, MAX(BusinessDate) BusinessDate
        FROM `p-cdh`.derivative.dexposure
        WHERE BusinessDate <= business_date
        GROUP BY ExposureKey
) A
INNER JOIN   `{env}-cdh`.derivative.dexposure B
on a.ExposureKey = b.ExposureKey
and a.BusinessDate = b.BusinessDate
AND B.IsActiveFlagInt = 0
AND coalesce(B.JustificationCode,0) != 43023

UNION

SELECT * FROM `{env}-cdh`.derivative.dexposure
WHERE BusinessDate = business_date
AND IsActiveFlagInt = 1
''')
```

#### get_collection_data

```python
env='p'
# Get the schema of the dexposure table
collection_table = spark.table(f"`{env}-cdh`.primary.collection")
collection_schema_fields = collection_table.schema.fields


# Convert the schema fields into the required string format
collection_schema = ",\n".join([
    f"{field.name} {field.dataType.simpleString()}"
    for field in collection_schema_fields
])


spark.sql(f'''
CREATE OR REPLACE FUNCTION `{env}-cdh`.derivative.get_collection_data(
    LastBusinessDate Date
)
RETURNS TABLE 
( 
{collection_schema}
)
RETURN

WITH date_info AS (
    SELECT
        c1.LastDateOfMonth,
        c1.PreviousMonthLastDate,
        c2.CalendarDate AS SecondBusinessDayNextReportingMonth,
        c1.FirstDateOfMonth,
        c3.CalendarDate as SecondBusinessDayOfMonth
    FROM `p-cdh`.reference.ref_calendar c1
    INNER JOIN `p-cdh`.reference.ref_calendar c2
        ON c2.BusinessDayOfMonth = 2
        AND c2.PreviousMonthLastWorkingDate = LastBusinessDate
    INNER JOIN `p-cdh`.reference.ref_calendar c3
        ON c3.BusinessDayOfMonth = 2
        AND c3.CurrentMonthLastWorkingDate = LastBusinessDate
    WHERE c1.CalendarDate = LastBusinessDate
)
 
SELECT collection.*
FROM `p-cdh`.`primary`.collection collection
JOIN date_info date_info
    ON  collection.BusinessDate = (SELECT MAX(BusinessDate) FROM `p-cdh`.`primary`.collection)
WHERE
    (
     collection.ActualDate <= date_info.LastDateOfMonth AND 
     collection.MatchedDate BETWEEN date_info.FirstDateOfMonth  AND date_info.SecondBusinessDayNextReportingMonth
    )
    AND NOT 
    (
      collection.ActualDate <= date_info.PreviousMonthLastDate
      AND collection.MatchedDate BETWEEN date_info.FirstDateOfMonth AND date_info.SecondBusinessDayOfMonth
    )
AND collection.FinalReason in ('Matched', 'matched')
''')
```
