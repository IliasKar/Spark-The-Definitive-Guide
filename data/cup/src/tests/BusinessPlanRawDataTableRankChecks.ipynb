##### Question: Why the two tables: **sme_business_plan** & **sme_business_plan_strategy** are join in the application_sme table with **INNER JOIN**?

```python
%sql

WITH sme_business_plan AS (
SELECT *
FROM `p-cdh`.landed.sme_business_plan
),

sme_business_plan_strategy AS (
SELECT ID AS id_strategy,
       BusinessPlanID AS BusinessPlanID_strategy, *
FROM `p-cdh`.landed.sme_business_plan_strategy
),

strategies_after_inner_join AS (
SELECT x.BusinessDate, 
       COUNT(x.ID) AS CountStrategiesAfterInnerJoin,
       COUNT(DISTINCT x.ID) AS CountDistinctStrategiesAfterInnerJoin
FROM sme_business_plan AS x
INNER JOIN sme_business_plan_strategy AS y
ON x.BusinessDate = y.BusinessDate
AND x.ID = y.id_strategy
GROUP BY x.BusinessDate
),

business_plan_strategies AS (
SELECT BusinessDate, -- 511 Distinct BusinessDates 
       COUNT(ID) AS CountStrategiesBeforeInnerJoin,
       COUNT(DISTINCT ID) AS CountDistinctStrategiesBeforeInnerJoin
FROM sme_business_plan
GROUP BY BusinessDate
)

SELECT COUNT(*) AS CountProblematicRows
FROM (
       SELECT x.*, 
              y.CountStrategiesAfterInnerJoin,
              y.CountDistinctStrategiesAfterInnerJoin
       FROM business_plan_strategies AS x
       LEFT JOIN strategies_after_inner_join AS y
       ON x.businessdate  = y.businessdate
       )
WHERE CountStrategiesBeforeInnerJoin <> CountStrategiesAfterInnerJoin


-- Intial rows in the table: 6,437
-- After INNER JOIN: 6,437 - THEREFORE, WE ARE NOT FILTERING OUT ANY BUSINESS STRATEGY!!
```
