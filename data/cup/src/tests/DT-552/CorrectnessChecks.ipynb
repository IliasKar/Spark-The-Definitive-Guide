```python
%sql
WITH real_estate_building_after_correction AS (
SELECT 'After correction' AS Status,
         COUNT(*) AS countRows
FROM `d-cdh`.primaryactive.RealEstateBuilding
WHERE BusinessDate = '2026-03-19'
AND SourceSystem = 'thesis_RES'
),

real_estate_building_before_correction AS (
SELECT 'Before correction' AS Status,
        COUNT(*) AS countRows
FROM `p-cdh`.primary.realestatebuilding
WHERE BusinessDate = '2026-03-19'
AND SourceSystem = 'thesis_RES'    
),

real_estate_building_after_correction_with_non_null_floors AS (
SELECT 'After correction - non null floors' AS Status,
        COUNT(*) AS countRows
FROM `d-cdh`.primaryactive.RealEstateBuilding
WHERE BusinessDate = '2026-03-19'
AND SourceSystem = 'thesis_RES'
AND RealEstateBuildingFloorRawValue IS NOT NULL  
),

real_estate_building_after_correction_with_non_null_net_area AS (
SELECT 'After correction - non null net area' AS Status,
        COUNT(*) AS countRows
FROM `d-cdh`.primaryactive.RealEstateBuilding
WHERE BusinessDate = '2026-03-19'
AND SourceSystem = 'thesis_RES'
AND RealEstateNetArea IS NOT NULL  
)

SELECT *
FROM real_estate_building_after_correction
UNION ALL
SELECT *
FROM real_estate_building_before_correction
UNION ALL
SELECT *
FROM real_estate_building_after_correction_with_non_null_floors 
UNION ALL
SELECT *
FROM real_estate_building_after_correction_with_non_null_net_area
```
