```python
def load_curated_product_arotron(
    spark, curated_zone, catalog_landed, reference_zone, run_date, schema, env_prefix
) -> None:
    args = [curated_zone, catalog_landed, reference_zone, run_date, schema, env_prefix]
    if all(item is not None for item in args) :
        date_filter = f"businessdate = {run_date!r}"

        # schema = (
        #     "ProductId  string "
        #     ",SourceSystem string"
        #     ",IngestionTimestamp timestamp"
        #     ",BatchId string"
        #     ",ProductKey  string "
        #     ",ProductCode  string "
        #     ",ProductDescription  string"
        #     ",SubProductCode  int"
        #     ",SubProduct  string"
        #     ",BankingUnitCode  string"
        #     ",InterestUnit  string"
        #     ",InterestFrequency  int"
        #     ",DeliquentInterestUnit  string"
        #     ",DeliquentInterestFrequency  int"
        #     ",InterestDate  date"
        #     ",LastValeurDate  date"
        #     ",NextValeurDate  date"
        #     ",PeriodicIntFlag  string"
        #     ",ShadowFlag  boolean"
        #     ",SyndicatedFlag  boolean"
        #     ",L3869Flag  boolean"
        #     ",OCWFlag  boolean"
        #     ",BondedFlag  boolean"
        #     ",ProductCategory  string"
        #     ",BusinessDate  date"
        # )

        cluster_columns = ["BusinessDate"]

        curated_table_path = f"{curated_zone}.Product"

        # ================== Create Delta Table ==================
        logging.info("Creating Delta Table...")
        if spark.catalog.tableExists(curated_table_path) is False:
            create_delta_table(
                spark,
                curated_table_path,
                cluster_columns,
                schema,  # type: ignore
            )
        else:
            logging.info("Table already exists.")

        # ================== Delete From Delta Table ==================
        if spark.catalog.tableExists(curated_table_path) is True:
            delete_from_curated_table(
                spark,
                curated_table_path,
                run_date,  # type: ignore
                "arotron",
            )        
        else:
            logging.info("Tables does not exist, no need to delete entries for business date!")

        # ========================== eaisig_be ==========================
        eaisig_be_extra_filters = None

        if eaisig_be_extra_filters is not None:
            eaisig_be_extra_filters.append(date_filter)
            eaisig_be_final_filters = " AND ".join(eaisig_be_extra_filters)
        else:
            eaisig_be_final_filters = date_filter

        eaisig_be_selected_fields = [
            "eaisig_be_sprd_s_id",
            "eaisig_be_prd_s_id",
            "businessdate",
        ]

        eaisig_be = load_landed_delta_table(
            spark,
            catalog_landed,  # type: ignore
            "arotron_eaisig_be",
            eaisig_be_final_filters,
            eaisig_be_selected_fields,
        )

        eaisig_be_primary_key_fields = [eaisig_be.eaisig_be_sprd_s_id, lit("arotron")]

        product_arotron = eaisig_be.withColumn(
            "ProductId", sha2(concat_ws(";", *eaisig_be_primary_key_fields), 256)
        )

        product_arotron = product_arotron.select(
            product_arotron.ProductId,
            lit("arotron").alias("SourceSystem"),
            lit(from_utc_timestamp(current_timestamp(), "Europe/Athens")).alias(
                "IngestionTimestamp"
            ),
            lit(None).alias("BatchId"),
            product_arotron.eaisig_be_sprd_s_id.cast("string").alias("ProductKey"),
            product_arotron.eaisig_be_sprd_s_id.cast("string").alias("ProductCode"),
            product_arotron.eaisig_be_prd_s_id.cast("string").alias(
                "ProductDescription"
            ),
            lit(None).cast("int").alias("SubProductCode"),
            lit(None).cast("string").alias("SubProduct"),
            lit(None).cast("string").alias("BankingUnitCode"),
            lit(None).cast("string").alias("InterestUnit"),
            lit(None).cast("int").alias("InterestFrequency"),
            lit(None).cast("string").alias("DeliquentInterestUnit"),
            lit(None).cast("int").alias("DeliquentInterestFrequency"),
            lit(None).cast("date").alias("InterestDate"),
            lit(None).cast("date").alias("LastValeurDate"),
            lit(None).cast("date").alias("NextValeurDate"),
            lit(None).cast("string").alias("PeriodicIntFlag"),
            lit(None).cast("boolean").alias("ShadowFlag"),
            lit(None).cast("boolean").alias("SyndicatedFlag"),
            lit(None).cast("boolean").alias("L3869Flag"),
            lit(None).cast("boolean").alias("OCWFlag"),
            lit(None).cast("boolean").alias("BondedFlag"),
            lit(None).cast("string").alias("ProductCategory"),
            product_arotron.businessdate.alias("BusinessDate"),
        ).distinct()  # granularity account_number distinct to eliminate

        # ================== Append ==================

        product_arotron.write.format("delta").clusterBy("BusinessDate").mode(
            "append"
        ).saveAsTable(curated_table_path)
    else:
        null_args = [
            arg_name
            for arg_name, arg in args
            if arg
            in [
                curated_zone,
                catalog_landed,
                reference_zone,
                run_date,
                schema,
                env_prefix,
            ]
            and arg is None
        ]

        for arg_name in null_args:
            logging.error(f"No {arg_name} specified!")

    logging.info("We did it!")
```

```python
load_curated_product_arotron(spark, curated_zone, catalog_landed, reference_zone, run_date, schema, env_prefix)
```
